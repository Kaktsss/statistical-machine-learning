## 5. Functions

Functions let you package reusable logic. Define with `def`, return values with `return`.

```python
# Basic function
def greet(name, greeting="Hello"):
    """Return a greeting string. `greeting` has a default value."""
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", greeting="Good morning"))
```

```python
# Function returning multiple values
def stats(data):
    """Return mean and standard deviation of a list."""
    n    = len(data)
    mean = sum(data) / n
    variance = sum((x - mean)**2 for x in data) / n
    std  = variance ** 0.5
    return mean, std          # returns a tuple

data = [12, 15, 14, 10, 18, 22, 9, 13]
mu, sigma = stats(data)      # unpack the tuple
print(f"Mean = {mu:.2f}, Std = {sigma:.2f}")
```

```python
# Lambda — anonymous one-liner function
square = lambda x: x ** 2
celsius_to_fahrenheit = lambda c: c * 9/5 + 32

print("3² =", square(3))
print("100 °C =", celsius_to_fahrenheit(100), "°F")
```

## 8. Fitting a Regression

We'll look at **two common approaches**:
- **`numpy.polyfit`** — simple polynomial (including linear) fitting
- **`scipy.stats.linregress`** — linear regression with full statistics

### 8.1 Linear Regression with NumPy

```python
# Generate noisy linear data
rng  = np.random.default_rng(0)
x    = np.linspace(0, 10, 50)
y    = 3.0 * x + 5.0 + rng.normal(0, 4, size=len(x))  # true slope=3, intercept=5

# Fit a degree-1 polynomial (i.e. a straight line) → returns [slope, intercept]
coeffs = np.polyfit(x, y, deg=1)
slope, intercept = coeffs
print(f"Fitted line:  y = {slope:.3f} · x + {intercept:.3f}")

# Build the fitted line
y_fit = np.polyval(coeffs, x)

# Plot
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(x, y, label="Data",  color="steelblue", edgecolors="white", s=50, zorder=3)
ax.plot(x, y_fit, label=f"Fit: y = {slope:.2f}x + {intercept:.2f}",
        color="tomato", linewidth=2.5)
ax.set_title("Linear Regression with NumPy")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 8.2 Linear Regression with SciPy — Full Statistics

```python
# before running this chunk you need to install numpy
# python3 -m pip install scipy

from scipy import stats

result = stats.linregress(x, y)

print("=== SciPy Linear Regression Results ===")
print(f"  Slope      : {result.slope:.4f}")
print(f"  Intercept  : {result.intercept:.4f}")
print(f"  R²         : {result.rvalue**2:.4f}")
print(f"  p-value    : {result.pvalue:.2e}")
print(f"  Std error  : {result.stderr:.4f}")
```

```python
# Confidence band using standard error
y_pred  = result.slope * x + result.intercept
margin  = 1.96 * result.stderr * np.sqrt(1/len(x) + (x - x.mean())**2 / ((x - x.mean())**2).sum())

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(x, y, color="steelblue", edgecolors="white", s=50, label="Data", zorder=3)
ax.plot(x, y_pred, color="tomato", linewidth=2.5,
        label=f"Fit: y = {result.slope:.2f}x + {result.intercept:.2f}  (R²={result.rvalue**2:.3f})")
ax.fill_between(x, y_pred - margin, y_pred + margin,
                color="tomato", alpha=0.15, label="95% Confidence Band")
ax.set_title("Linear Regression with Confidence Band")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### 8.3 Polynomial Regression

```python
# Generate non-linear data
rng = np.random.default_rng(7)
x_nl = np.linspace(-3, 3, 60)
y_nl = 0.5 * x_nl**3 - 2 * x_nl + rng.normal(0, 1.5, len(x_nl))

# Fit polynomials of different degrees
x_smooth = np.linspace(-3, 3, 300)

fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=True)

for ax, deg in zip(axes, [1, 2, 3]):
    coeffs  = np.polyfit(x_nl, y_nl, deg=deg)
    y_curve = np.polyval(coeffs, x_smooth)

    # R² manually
    y_hat = np.polyval(coeffs, x_nl)
    ss_res = np.sum((y_nl - y_hat)**2)
    ss_tot = np.sum((y_nl - y_nl.mean())**2)
    r2 = 1 - ss_res / ss_tot

    ax.scatter(x_nl, y_nl, color="steelblue", edgecolors="white", s=30, alpha=0.8)
    ax.plot(x_smooth, y_curve, color="tomato", linewidth=2.5)
    ax.set_title(f"Degree {deg}  (R²={r2:.3f})")
    ax.set_xlabel("x")
    ax.grid(alpha=0.3)

axes[0].set_ylabel("y")
fig.suptitle("Polynomial Regression — Comparing Degrees", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```
